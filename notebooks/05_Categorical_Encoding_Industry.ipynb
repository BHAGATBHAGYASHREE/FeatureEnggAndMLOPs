Machine learning algorithms require numerical input, so categorical variables must be converted into numbers without losing meaning.


```
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import LabelEncoder, OneHotEncoder, OrdinalEncoder

df=pd.read_csv("customer_categorical_encoding.csv")
df.head()
```

```
# Label Encoding (good for binary features)
le=LabelEncoder()
df['Gender_Label']=le.fit_transform(df['Gender'])
df[['Gender','Gender_Label']].head()
```

```
# Ordinal Encoding
ord_enc=OrdinalEncoder(categories=[['Basic','Silver','Gold','Platinum']])
df['Plan_Ordinal']=ord_enc.fit_transform(df[['SubscriptionPlan']])
df[['SubscriptionPlan','Plan_Ordinal']].head()
```

```
# One Hot Encoding
ohe=pd.get_dummies(df[['Country','DeviceType']],drop_first=False)
print(ohe.head())
print("New Columns:",len(ohe.columns))
```

# High Cardinality Problem

Imagine replacing **Country** with **PostalCode** containing 50,000 unique values.

One-Hot Encoding would create 50,000 columns.

### Industry Solutions

- Target Encoding
- Frequency Encoding
- Hash Encoding
- Embeddings (Deep Learning)

Use these techniques when cardinality is very high.


# Data Leakage Warning

Target Encoding calculates statistics from the target variable.

Never compute target encoding using the entire dataset before train-test split.

Always fit encoders on the training data only.

This prevents data leakage and unrealistic model performance.


```
plt.figure(figsize=(6,4))
df['SubscriptionPlan'].value_counts().plot(kind='bar')
plt.title('Subscription Plans')
plt.tight_layout()
plt.show()
```
